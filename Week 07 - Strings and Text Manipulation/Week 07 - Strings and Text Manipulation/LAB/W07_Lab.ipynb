## Topic 1 · Case and whitespace
*Walkthrough: `01 - Case and Whitespace`*

### D1. Case changes comparisons

1. Create `a = "SRV-01"` and `b = "srv-01"`.
2. Print `a == b`.
3. Print the same comparison again, with `.lower()` on both sides.

**Expected output:** `False`, then `True`

````python
# D1
a = "SRV-01"
b = "srv-01"

print(a == b) # False because uppercase and lowercase differ (case-sensitive)
print(a.lower() == b.lower()) # True because both are converted to lowercase so they match (case-insensitive)
````

### D2. `.upper()`, `.lower()`, `.title()`

1. Create `name = "record check console"` and print it with `.title()`, then with `.upper()`.
2. Create `original = "SRV-01"`. Print `original.lower()`, then print `original` again.
3. Add a comment explaining why `original` did not change.

**Expected output:**
```
Record Check Console
RECORD CHECK CONSOLE
srv-01
SRV-01
```

````python
# D2
name = "record check console"
print(name.title()) # title() capitalizes each word
print(name.upper()) # upper() converts all letters to uppercase

original = "SRV-01"
print(original.lower()) # lower() converts all letters to lowercase
print(original) # original is unchanged because we did not reassign it
````

### D3. `.strip()`

1. Create `raw = "  srv-01  "`.
2. Print it inside square brackets so you can see the spaces: `print(f"[{raw}]")`
3. Print it again the same way, with `.strip()`.

**Expected output:** `[  srv-01  ]`, then `[srv-01]`

````python
# D3
raw = " srv-01 "
print(f"[{raw}]") # brackets make the surrounding spaces visible
print(f"[{raw.strip()}]") # .strip() removes whitespaces from both sides
````

### D4. `.lstrip()`, `.rstrip()` and chaining

1. Create `messy = "***srv-01***"` and print `messy.strip("*")`, `messy.lstrip("*")` and `messy.rstrip("*")`.
2. Create `field = "  SRV-01  "` and print `field.strip().lower()` (two methods chained together).

**Expected output:** `srv-01`, `srv-01***`, `***srv-01`, then `srv-01`

````python
# D4
messy = "***srv-01***"
print(messy.strip("*")) # removes stars from both sides
print(messy.lstrip("*")) # removes stars only from the left side
print(messy.rstrip("*")) # removes stars only from the right side

field = " SRV-01"
print(field.strip().lower()) # removes surrounding whitespace and converts it to lowercase
````

### D8. `.startswith()`, `.endswith()`, `.count()`

### D13. `.join()` needs strings, and f-strings build output

## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
raw_line = " Srv-01, used=87 , total=100 "
line = raw_line.strip() # removes whitespace from the sides of the entire line
hostname_part, rest = line.split(",", 1) # separates the hostname from the fields
hostname = hostname_part.strip().lower() # cleans the hostname by removing whitespace and converting to lowercase

# stores each cleaned key with its numerical value
fields = {}
for pair in rest.split(","):
    key, value = pair.split("=")
    fields[key.strip()] = float(value.strip())

# combines the hostname and numerical fields into one dictionary
field = {"hostname": hostname,
         "used": fields["used"],
         "total": fields["total"]
    }
print(field)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """Turn a messy server line into a clean dictionary."""
    line = raw_line.strip() # removes whitespace from the sides of the entire line
    hostname_part, rest = line.split(",", 1) # separates the hostname from the fields
    hostname = hostname_part.strip().lower() # cleans the hostname by removing whitespace and converting to lowercase

    # stores each cleaned key with its numerical value
    fields = {}
    for pair in rest.split(","):
        key, value = pair.split("=")
        fields[key.strip()] = float(value.strip())
    # returns one dictionary containing the hostname and numerical fields
    return {"hostname": hostname,
         "used": fields["used"],
         "total": fields["total"]
    }
print(parse_line("SRV-02,used=120,total=100"))
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
# Reuse parse_line() from D15

# Return a status based on the percentage used.
def status_of(used,total):
    percent = (used / total) * 100
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"

# reads each line, parses it, and prints its hostname and status
with open("messy_lines.txt") as f:
    for raw_line in f:
        clean = parse_line(raw_line)
        status = status_of(clean["used"], clean["total"])
        print(clean["hostname"], status)
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**
```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}
Could not parse: srv-08,used=unknown,total=100
Could not parse: srv-09,note=a=b,used=1,total=2
```

````python
# C2
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]

for line in lines:
    try: 
        print(parse_line(line)) # converts the line into a clean dictionary and prints it
    except ValueError:
        print("Could not parse:", line) # prints a message along with the line that failed without stopping the loop
````
