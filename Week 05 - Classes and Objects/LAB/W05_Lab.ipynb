### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"
    # calculate the percentage 
    def status(self):
        p = (self.value / self.limit) * 100
        # return the status based on the percentage
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else: 
            return "OK"

# create three Record objects with different values
r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)
# print each record's status
print(r1.status())
print(r2.status())
print(r3.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        # return the value as a percentage of limit
        return (self.value / self.limit) * 100
    
    def status(self):
        # get the percentage by calling the percent method
        p = self.percent()
        # check the conditions and return the status based on the percentage
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else: 
            return "OK"
        
    
# create a Record object and print its percentage and status
record = Record("R-001", 87.0, 100.0)
print(record.percent())
print(record.status())
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    def __init__(self):
        # start the object's count at 0
        self.over_limit_count = 0

    def check(self, value, limit):
        # add 1 if the value is greater than the limit
        if value > limit:
            self.over_limit_count += 1

# create one Checker instance
checker = Checker()

# call the method three times with different values
checker.check(87, 100)
checker.check(120, 100)
checker.check(130, 100)
# print how many checks are oer the limit 
print(checker.over_limit_count)
````

### D9. A list of objects

1. Using your `Record` class from D7, create a list of three records:
   `R-001` (87.0), `R-002` (120.0) and `R-003` (45.0), each with a limit of 100.0.
2. Loop over the list and print each record's label and status.

**Expected output:**
```
R-001 OK
R-002 OVER LIMIT
R-003 OK
```

````python
# D9
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        # return the value as a percentage of limit
        return (self.value / self.limit) * 100
    
    def status(self):
        # get the percentage by calling the percent method
        p = self.percent()
        # check the conditions and return the status based on the percentage
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else: 
            return "OK"
# create a list of three Record objects       
records = [
    Record("R-001", 87.0, 100.0),
    Record("R-002", 120.0, 100.0),
    Record("R-003", 45.0, 100.0),
]
# loop through the records and print each label and status of those records
for record in records:
    print(record.label, record.status())
````

### D10. A class-level counter

1. Define a new, small class `Counted`. Directly under the `class` line (not inside `__init__`),
   add `count = 0`.
2. In `__init__(self)`, add 1 to `Counted.count`.
3. Create three `Counted` objects, then print `Counted.count`.

**Expected output:** `3`

Add a comment explaining why the count is shared by every object instead of each object having its own.

````python
# D10
# Explanation: the count is shared to keep track of the total number of objects created.
# if each object had its own count, then it would only count itself.

class Counted:
    # the count belongs to the class, so all the objects share it, also starting count is 0
    count = 0
    def __init__(self):
        # increase the shared count by 1 whenever an object is created
        Counted.count += 1

# create three Counted objects
obj1 = Counted()
obj2 = Counted()
obj3 = Counted()
# print the total number of objects created
print(Counted.count)
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path):
    # start with an empty list
    records = []
    # open the file and read each line
    with open(path) as f:
        for line in f:
            # remove the whitespace and separate the values at each comma
            label, value, limit = line.strip().split(",")
            # conver the numbers into floats and add the record to the list
            records.append(Record(label, float(value), float(limit)))
    # return the list of records
    return records
# load the records from the "records_sample.txt" file 
records = load_records("records_sample.txt")
# pritn each record's label and status
for record in records:
    print(record.label, record.status())
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit
    # return True if the value is greater than the limit
    def is_over(self):
        return self.value > self.limit

def load_records(path):
    # start with an empty list
    records = []
    # open the file and read each line
    with open(path) as f:
        for line in f:
            # remove the whitespace and separate the values at each comma
            label, value, limit = line.strip().split(",")
            # conver the numbers into floats and add the record to the list
            records.append(Record(label, float(value), float(limit)))
    # return the list of records
    return records
# load the records from the "record_sample.txt" file
records = load_records("records_sample.txt")
# start the count at 0
over_limit_count = 0
# if the record is over limit, then add 1 to the count 
for record in records:
    if record.is_over():
        over_limit_count += 1
# print the number over limit and the total number of records
print(f"{over_limit_count} of {len(records)} records are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self, username):
        # store the username and set this account's failed login count to 0 
        self.username = username
        self.failed_logins = 0

    def fail(self):
        # add 1 to this account's failed login count
        self.failed_logins += 1
    # return True if there have been 3 or more failed logins
    def locked(self):
        return self.failed_logins >= 3
# create Anna's account and call the method fail three times
anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
# check if Anna's account is locked or not
print(anna.locked())

# create Ben's account and call the method fail only once
ben = Account("ben")
ben.fail()
# check if Ben's account is locked or not
print(ben.locked())
````
