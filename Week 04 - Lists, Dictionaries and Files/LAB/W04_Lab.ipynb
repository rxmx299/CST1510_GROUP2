### D1. Create a list and index into it

1. Create `labels = ["R-001", "R-002", "R-003"]`.
2. Print the first item with `labels[0]`, the last item with a **negative** index, and the length with `len()`.

**Expected output:** `R-001`, `R-003`, `3`

### D2. Slicing

1. Create `values = [5, 10, 15, 20, 25, 30]`.
2. Use slicing to print: the middle four, the first two, and everything from position 3 to the end.

**Expected output:**
```
[10, 15, 20, 25]
[5, 10]
[20, 25, 30]
```

*Remember: the stop position is not included.*

````python
# D2
values = [5, 10, 15, 20, 25, 30]
print(values[1:5]) # middle four: [10, 15, 20, 25]
print(values[:2]) # first two: [5,10]
print(values[3:]) # from position 3 to the end: [20, 25, 30]
````

### D3. Change a list, then loop over it

1. Create `values = [10, 20, 30]`.
2. Change the first item to `99` and print the list.
3. Add `40` to the end with `.append()` and print the list.

**Expected output:**

````python
# D3
values = [10, 20, 30]
# change the first item to 99
values[0] = 99
print(values)
# add 40 to the end
values.append(40)
print(values)
# remove the value 20
values.remove(20)
print(values)
# print each value from the updated list on its own line
for value in values:
    print(value)
````

### D6. List comprehensions

1. Create `values = [40, 55, 70, 25]`.
2. Using a list comprehension, make a new list with every value doubled, and print it.
3. Using a list comprehension with `if`, make a new list with only the values that are 50 or more, and print it.

**Expected output:**
```
[80, 110, 140, 50]
```

````python
# D6
values = [40, 55, 70, 25]

# create a new list with each value doubled
doubled = [value * 2 for value in values]
print(doubled)

# create a new list with only the values that are 50 or more
big_values = [value for value in values if value >= 50]
print(big_values)
````

## Topic 2 · Dictionaries and sets
*Walkthrough: `02 - Dictionaries`*

### D7. Create, read, add and update

1. Create `record = {"label": "R-001", "value": 87.0, "limit": 100.0}`.
2. Print the label using its key.
3. Add a new key `"status"` with the value `"OK"`.
4. Change `"value"` to `90.0`.
5. Print the whole dict.

**Expected output:**
```
R-001
{'label': 'R-001', 'value': 90.0, 'limit': 100.0, 'status': 'OK'}
```

````python
# D7
record = {"label": "R-001", "value": 87.0, "limit": 100.0}

# print the label using its key
print(record["label"])

# add a new key called 'status'
record["status"] = "OK"

# change/update the value stored in the key 'value' to 90.0
record["value"] = 90.0

# print the whole dictionary
print(record)
````

### D8. Missing keys and `.get()`

Using `record` from D7:

1. Print `record.get("status")`.
2. Print `record.get("owner")`: this key does not exist.
3. Print `record.get("owner", "n/a")`.

**Expected output:** `OK`, `None`, `n/a`

````python
# D8
print(record.get("status")) # print the value of status becasse it exists
print(record.get("owner")) # print None because 'owner' does not exist
print(record.get("owner","n/a")) # print "n/a" as a fallback because 'owner' does not exist
````

### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```
label: R-001
value: 90.0
limit: 100.0
status: OK
```

````python
# D9
# print each key in the dictionary
for key in record:
    print(key)

# print each value in the dictionary
for value in record.values():
    print(value)

# print each key with its value
for key,value in record.items():
    print(f"{key}: {value}")
````

### D10. Build a dict from two lists with `zip()`

1. Create `hostnames = ["srv-01", "srv-02", "srv-03"]` and `used = [87, 45, 120]`.
2. Build a dict with `dict(zip(hostnames, used))` and print it.
3. Print the value for `"srv-03"`.

**Expected output:**
```
{'srv-01': 87, 'srv-02': 45, 'srv-03': 120}
120
```

````python
# D10
hostnames = ["srv-01","srv-02","srv-03"]
used = [87, 45, 120]

# pair each hostname with its used value and create dictionary
usage = dict(zip(hostnames, used))
print(usage)

# print the value stored under "srv-03"
print(usage["srv-03"])
````

### D11. Sets: unique values and quick checks

1. Create `ips = ["10.0.0.5", "10.0.0.9", "10.0.0.5", "10.0.0.7"]`.
2. Make a set from the list, and print how many **unique** IPs there are.
3. Print whether `"10.0.0.9"` is in the set, and whether `"10.0.0.1"` is.

**Expected output:** `3`, `True`, `False`

````python
# D11
ips = ["10.0.0.5", "10.0.0.9", "10.0.0.5", "10.0.0.7"]

# create a set to keep each ip address only once (remove duplicates)
unique_ips = set(ips)

# print the number of unique ip addresses
print(len(unique_ips))

# check whether each address is in the set
print("10.0.0.9" in unique_ips)
print("10.0.0.1" in unique_ips)
````

### D13. Read a file line by line and split each line

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

## Topic 4 · Handling errors
*Walkthrough: `04 - Handling Errors`*

### D14. Catch a `ValueError`

1. Create `text = "abc"`.
2. Inside `try:`, convert it with `float(text)` and print the number.
3. In `except ValueError:`, print `Not a number: abc` instead.
4. Now change `text` to `"42"` and run again.

**Expected output:** `Not a number: abc`, then (after the change) `42.0`

````python
# D14
text = "42"

try:
    # try to convert the text to a number and then print it
    number = float(text)
    print(number)
except ValueError:
    # print this message if the text cannot be converted (instead of an error crashing the program)
    print(f"Not a number: {text}")
````

### D15. Catch a `FileNotFoundError`

1. Inside `try:`, open `missing.txt` with `with` and print its contents.
2. In `except FileNotFoundError:`, print `missing.txt not found`.

**Expected output:** `missing.txt not found`

````python
# D15
try:
    # try to open the file,read its contents and print them.
    with open("missing.txt") as f:
        print(f.read())
except FileNotFoundError:
    # print this message if the file does not exist
    print("missing.txt not found")
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """Reads records from a file, and returns an empty list if the file is missing."""
    records = []

    try:
        with open(path) as f:
            for line in f:
                # remove the newline and separate the values at each comma
                label, value, limit = line.strip().split(",")
                # create a dictionary for this record and add it to the list
                records.append({
                    "label": label,
                    "value": float(value),
                    "limit": float(limit)
                })
    except FileNotFoundError:
        print(f"File not found: {path}")
        return []

    return records

# load the records and print each label with its status
records = load_records("records_sample.txt")

for record in records:
    status = "OVER LIMIT" if record["value"] > record["limit"] else "OK"
    print(f"{record['label']}: {status}")

# Test a missing file without replacing the records loaded above
print(load_records("nope.txt"))
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2
# write each label and status to results.txt
with open("results.txt", "w") as f:
    for record in records:
        status = "OVER LIMIT" if record["value"] > record["limit"] else "OK"
        f.write(f"{record['label']},{status}\n")

# read the results back and print them
with open("results.txt") as f:
    print(f.read())
````
