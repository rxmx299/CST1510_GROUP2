## Topic 1 · Series and DataFrame
*Walkthrough: `01 - Series and DataFrame`*

### D1. A Series

1. Run `import pandas as pd`.
2. Create `values = pd.Series([87, 120, 45])` and print it.
3. Print `values.max()`.

**Expected output:**
```
0     87
1    120
2     45
dtype: int64
120
```

````python
# D1
import pandas as pd # importing pandas library for working with dataframes and series

values = pd.Series([87, 120, 45]) # creates a series that contains three values
print(values) # displays the series
print(values.max()) # displays the largest value in the series
````

### D2. A DataFrame from a dict

1. Create a DataFrame from a dict of lists, with three columns:
   `label` (`"R-001"`, `"R-002"`, `"R-003"`), `value` (`87`, `120`, `45`) and `limit` (`100`, `100`, `100`).
2. Print it, then print its `.shape`.

**Expected output:** a table with 3 rows and 3 columns, then `(3, 3)`

````python
# D2
# creates a dataframe from a dictionary of lists
df = pd.DataFrame({
    "label": ["R-001", "R-002", "R-003"],
    "value": [87, 120, 45],
    "limit": [100, 100, 100]
})
print(df) # displays the dataframe
print(df.shape) # displays the number of rows and columns in the dataframe
````

### D3. Read a whole file with `read_csv()`

1. Read `records_sample.csv` (in this `LAB` folder) into a DataFrame called `df`.
2. Print `df`, then print `df.dtypes`.

**Expected output:** a table with 6 records (`R-001` to `R-006`), then the types:
`value` and `limit` are `int64`, and `label` is text (shown as `object`, or `str` in newer versions of pandas).

````python
# D3
df = pd.read_csv("records_sample.csv") # reada the CSV file into a dataframe
print(df) # displays the dataframe
print(df.dtypes) # displays the data types of each column in the dataframe
````

### D4. Add a column, without a loop

Add a `percent` column to `df`: `(value / limit) * 100`. Print `df`.

**Expected output:** the `percent` column holds `87.0`, `120.0`, `45.0`, `93.0`, `101.0`, `60.0`.

````python
# D4
df["percent"] = (df["value"] / df["limit"]) * 100 # calculates the percentage of each value and adds it as a new column in the dataframe
print(df) # displays the updated dataframe with the new percent column
````

## Topic 2 · Selecting and filtering
*Walkthrough: `02 - Selecting and Filtering`*

### D5. `iloc` and `loc`

1. Print the `label` of the first row, using `df.iloc[0]["label"]`.
2. Print the `value` in the row with index `2`, using `df.loc[2, "value"]`.

**Expected output:** `R-001`, then `45`

````python
# D5
print(df.iloc[0]["label"]) # used iloc to select the first row and print its label value
print(df.loc[2, "value"]) # used loc to select the value in the third row and print it
````

### D6. Selecting columns

1. Print just the `label` column (one pair of brackets).
2. Print just the `label` and `percent` columns (a list inside the brackets).

**Expected output:** first one column of labels, then a two-column table.

````python
# D6
print(df["label"]) # displays the label column of the dataframe (as a series)
print(df[["label", "percent"]]) # displays the label and percent columns of the dataframe (as a dataframe)
````

### D7. Filter rows with a condition

1. Keep only the rows where `percent` is 100 or more, using a condition inside `df[...]`. Print them.
2. Print how many rows that is, with `len()`.

**Expected output:** the rows for `R-002` and `R-005`, then `2`

````python
# D7
over_limit = df[df["percent"] >= 100] # keeps only the records with a percentage of 100 or more

print(over_limit) # displays the records that are over the limit
print(len(over_limit)) # displays the number of records that are over the limit
````

### D8. Combine conditions with `&` and `|`

1. Print the rows where `percent` is 90 or more **and** less than 100.
2. Print the rows where `percent` is less than 50 **or** more than 110.

*Put brackets round each condition, and use `&` and `|`, not `and` / `or`.*

**Expected output:** first only `R-004`, then `R-002` and `R-003`

````python
# D8
# displays the records that are between 90% and 100% of the limit, using & to select the records where both conditions are true
print(df[(df["percent"] >= 90) & (df["percent"] < 100)])

# displays the record that are either below 50% OR above 110%, using | to select the records where either condition is true
print(df[(df["percent"] < 50) | (df["percent"] > 110)])
````

## Topic 3 · Grouping, sorting and aggregating
*Walkthrough: `03 - Grouping, Sorting and Aggregating`*

### D9. Sort

Sort `df` by `percent`, highest first, and print it.

**Expected output:** the labels in this order: `R-002`, `R-005`, `R-004`, `R-001`, `R-006`, `R-003`

````python
# D9
print(df.sort_values("percent", ascending=False)) # sorts the dataframe by the percent column in descending order and displays it
````

### D10. `groupby()`, `.agg()` and `value_counts()`

1. Add a column `over` that is `True` when `percent` is 100 or more: `df["over"] = df["percent"] >= 100`
2. Group by `over` and print the mean, max and min of `percent` for each group, using `.agg(["mean", "max", "min"])`.
3. Print `df["over"].value_counts()`.

**Expected output:**
```
         mean    max    min
over
False   71.25   93.0   45.0
True   110.50  120.0  101.0
```
then `False` 4 and `True` 2.

````python
# D10
df["over"] = df["percent"] >= 100 # creates a new column called "over" that contains True if the percent is greater than or equal to 100, and False otherwise
grouped = df.groupby("over") # groups the dataframe by the "over" column according to whether tey are over the limit or not
print(grouped["percent"].agg(["mean", "max", "min"])) # displays the mean, max, and min of the percent column for each group (over the limit or not)
print(df["over"].value_counts()) # displays the number of records that are over the limit and the number of records that are not over the limit
````

## Topic 4 · Apply, lambda and finishing up
*Walkthrough: `04 - Apply, Lambda and Finishing Up`*

### D11. A lambda

1. Write `double = lambda x: x * 2` and print `double(21)`.
2. Underneath, write the same thing as a normal `def` function called `double_def`, and print `double_def(21)`.

**Expected output:** `42`, then `42`

````python
# D11
double = lambda x: x * 2 # creates a lambda function that takes a single argument x and returns x multiplied by 2
print(double(21))

def double_def(x): # defines a function called double_def that performs the same calculation
    return x * 2

print(double_def(21))
````

### D12. `.apply()` with a lambda

1. Add a `status` column to `df` using `df["percent"].apply(lambda p: ...)`:
   `"OVER LIMIT"` if 100 or more, `"WARNING"` if 90 or more, otherwise `"OK"`.
2. Print just the `label` and `status` columns.
3. Print `df["status"].value_counts()`.

**Expected output:** `R-001` OK, `R-002` OVER LIMIT, `R-003` OK, `R-004` WARNING, `R-005` OVER LIMIT,
`R-006` OK; then `OK` 3, `OVER LIMIT` 2, `WARNING` 1.

````python
# D12
# created a new column called "status" and used a lambda function to assign a status based on each percentage value
df["status"] = df["percent"].apply(lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK")

print(df[["label", "status"]]) # displays each record's label and assigned status
print(df["status"].value_counts()) # displays the number of records that are assigned to each status category
````

### D13. Summarise and save

1. Print `df.describe()`.
2. Save `df` to a new file with `df.to_csv("records_report.csv", index=False)`.

**Expected output:** a summary table (count, mean, min, max and so on) for the number columns, and a new file
`records_report.csv` in this `LAB` folder.

````python
# D13
print(df.describe()) # displays summary statistics for the numeric columns in the dataframe

df.to_csv("records_report.csv", index=False) # saves the full dataframe to a CSV file without the row index
````

### D14. A first chart

1. Run `import matplotlib.pyplot as plt`.
2. Make a bar chart of the status counts: `df["status"].value_counts().plot(kind="bar")`
3. Give it a title with `plt.title(...)`, save it with `plt.savefig("status_counts.png")`, then `plt.show()`.

**Expected output:** a bar chart with three bars (OK, OVER LIMIT, WARNING), and a new file `status_counts.png`.

````python
# D14
import matplotlib.pyplot as plt # importing the matplotlib library for data visualisations

df["status"].value_counts().plot(kind="bar") # creates a bar chart of the counts of each status category
plt.title("Status Counts") # adds a title to the chart
plt.savefig("status_counts.png") # saves the chart as a PNG file
plt.show() # displays the chart 
````

## Challenge · for a first

*These combine several ideas from this week. There are no step-by-step instructions.*

### C1. Load, filter, sort and save

Starting again from `records_sample.csv`: add `percent` and `status`, keep only the `OVER LIMIT` rows,
sort them by `percent` (highest first), and save the result to `over_limit.csv`. Print the result too.

**Expected output:** two rows, `R-002` (120.0) then `R-005` (101.0)

````python
# C1
records = pd.read_csv("records_sample.csv") # reads the CSV file into a dataframe 

records["percent"] = records["value"] / records["limit"] * 100 # adds a new column in the dataframe that calculates the percentage of each value 

# adds a new column in the dataframe and uses a lambda function to assign a status to each record based on its percentage value 
records["status"] = records["percent"].apply(lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK")

# keeps only the records that are over the limit and sorts them by percentage in descending order
over_limit = records[records["status"] == "OVER LIMIT"].sort_values("percent", ascending=False) 
over_limit.to_csv("over_limit.csv", index=False) # saves the records that are over the limit to a new CSV file without the row index
print(over_limit) # displays the filtered and sorted records 
````

### C2. Average per status

Using your DataFrame with the `status` column, print the average `percent` for each status,
rounded to 1 decimal place.

**Expected output:**
```
status
OK             64.0
OVER LIMIT    110.5
WARNING        93.0
```

````python
# C2
grouped = df.groupby("status") # groups the dataframe by status
print(grouped["percent"].mean().round(1)) # displays the average percentage for each status category and rounds the results to one decimal place
````
